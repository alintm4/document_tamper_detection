### Importing Datasets

```python
import tensorflow as tf 
from tensorflow import keras
```

```python
import os
from pathlib import Path

# Set base path dynamically for Jupyter Notebook
BASE_PATH = Path.cwd().parent / 'split_dataset'
print(BASE_PATH)
def load_image_dataset(directory, batch_size=32, image_size=(256, 256)):
    """
    Load image dataset from directory.
    
    Args:
        directory: Path to the dataset directory
        batch_size: Batch size for the dataset (default: 32)
        image_size: Target image size as tuple (height, width) (default: (256, 256))
    
    Returns:
        keras dataset or None if directory doesn't exist
    """
    if not os.path.exists(directory):
        print(f"Directory not found: {directory}")
        return None
    
    return keras.utils.image_dataset_from_directory(
        directory=directory,
        labels='inferred',
        label_mode='int',
        batch_size=batch_size,
        image_size=image_size
    )
```

```python
# Load datasets using the function with dynamic paths
train_au_ds = load_image_dataset(str(BASE_PATH / 'train' / 'Au'))
train_tp_ds = load_image_dataset(str(BASE_PATH / 'train' / 'Tp'))
train_mask_ds = load_image_dataset(str(BASE_PATH / 'train' / 'Mask'))

test_au_ds = load_image_dataset(str(BASE_PATH / 'test' / 'Au'))
test_tp_ds = load_image_dataset(str(BASE_PATH / 'test' / 'Tp'))
test_mask_ds = load_image_dataset(str(BASE_PATH / 'test' / 'Mask'))

validation_au_ds = load_image_dataset(str(BASE_PATH / 'val' / 'Au'))
validation_tp_ds = load_image_dataset(str(BASE_PATH / 'val' / 'Tp'))
validation_mask_ds = load_image_dataset(str(BASE_PATH / 'val' / 'Mask'))
```
